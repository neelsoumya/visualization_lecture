## 7. Key Takeaways and Best Practices

**Important Principles:**

1. **Choose the Right Chart Type**: Match your visualization to your data and message
   - Use **line plots** for trends over time
   - Use **scatter plots** to show relationships between variables
   - Use **bar charts** to compare categories
   - Use **histograms** to show distributions

2. **Label Everything**: Always include titles, axis labels, and legends
   - Clear labels help viewers understand your data
   - Use descriptive, concise language

3. **Use Color Wisely**: 
   - Limit the number of colors to avoid confusion
   - Use colorblind-friendly palettes (viridis, colorblind, etc.)
   - Ensure sufficient contrast between colors

4. **Manage Transparency (Alpha)**:
   - Use alpha values to reveal overlapping data points
   - Typical values: 0.5-0.7 for overlapping elements

5. **Scaling and Normalization**:
   - Use secondary y-axes when plotting variables with different scales
   - Use density normalization to compare distributions with different sample sizes

6. **White Space and Layout**:
   - Use `tight_layout()` to prevent overlapping labels
   - Don't overcrowd visualizations—consider subplots instead

7. **Common Pitfalls to Avoid**:
   - Pie charts (use bar charts instead for clarity)
   - 3D charts (often harder to read than 2D)
   - Too many variables at once (simplify!)
   - Misleading axis scales (start at 0 for most comparisons)
