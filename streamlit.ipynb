# Lesson on Streamlit


## What is Streamlit?

Streamlit is an open-source Python library that allows data scientists and machine learning engineers to create interactive web applications for their data projects with minimal effort. It focuses on simplicity and speed, enabling you to turn data scripts into shareable web apps in just a few lines of code, without needing extensive web development knowledge.

### Key Features:

*   **Python-centric:** Write everything in Python, from data processing to UI components.
*   **Fast Development:** Build and deploy apps quickly with its intuitive API.
*   **Interactive Widgets:** Easily add sliders, buttons, text inputs, and more.
*   **Live Updates:** Changes in your code are immediately reflected in the app.
*   **Built for Data Science:** Perfect for showcasing models, dashboards, and exploratory data analysis.

### Why use Streamlit?

*   **Prototype quickly:** Rapidly iterate on data apps.
*   **Share insights:** Make your data projects accessible to non-technical stakeholders.
*   **Interactive visualization:** Create dynamic plots and tables.
*   **Demonstrate ML models:** Build user interfaces for your machine learning predictions.

# More advanced code

```python
import streamlit as st
import pandas as pd
import numpy as np

st.write("Streamlit supports a wide range of data visualizations")

all_users = ["Alice", "Bob", "Charly"]
with st.container(border=True):
    users = st.multiselect("Users", all_users, default=all_users)
    rolling_average = st.toggle("Rolling average")

np.random.seed(19)
data = pd.DataFrame(np.random.randn(20, len(users)), columns=users)

if rolling_average:
    data = data.rolling(7).mean().dropna()

tab1, tab2 = st.tabs(["Chart", "Dataframe"])
tab1.line_chart(data, height=250)
tab2.dataframe(data, height=250, use_container_width=True)
```

```python
%%writefile advanced_app.py

import streamlit as st
import pandas as pd
import numpy as np

st.write("Streamlit supports a wide range of data visualizations")

all_users = ["Alice", "Bob", "Charly"]
with st.container(border=True):
    users = st.multiselect("Users", all_users, default=all_users)
    rolling_average = st.toggle("Rolling average")

np.random.seed(19)
data = pd.DataFrame(np.random.randn(20, len(users)), columns=users)

if rolling_average:
    data = data.rolling(7).mean().dropna()

tab1, tab2 = st.tabs(["Chart", "Dataframe"])
tab1.line_chart(data, height=250)
tab2.dataframe(data, height=250, use_container_width=True)
```

Now that you've saved the advanced Streamlit code to `advanced_app.py`, you can run it using the same `ngrok` setup. Remember to uncomment and set your `ngrok` authtoken in the cell below if you haven't already. Then, replace `app.py` with `advanced_app.py` in the `!streamlit run` command.

To save your `ngrok` authtoken securely in Google Colab, you should use Colab's **Secrets** feature. This allows you to store sensitive information like API keys or tokens without embedding them directly in your code, which is important for security and when sharing notebooks.

Here's how to do it:

1.  **Open the Secrets Panel:** In the left sidebar of your Colab notebook, click on the **"🔑 Secrets"** icon (it looks like a key).
2.  **Add a new secret:** Click the **"+ Add new secret"** button.
3.  **Name the secret:** For your `ngrok` authtoken, you could name it `YOUR_AUTHTOKEN` (or any other descriptive name you prefer, but remember it for later).
4.  **Enter the secret value:** Paste your `ngrok` authtoken into the "Value" field.
5.  **Save:** Click "Save secret".

Once saved, you can access this secret in your Python code using `from google.colab import userdata` and then `userdata.get('YOUR_AUTHTOKEN')` (replacing `'YOUR_AUTHTOKEN'` with the name you gave your secret).

```python
from pyngrok import ngrok

# Terminate any existing ngrok tunnels
ngrok.kill()

# Start ngrok tunnel for Streamlit on port 8501 (default Streamlit port)
# Replace 'YOUR_AUTHTOKEN' with your actual ngrok authtoken if prompted by ngrok

from google.colab import userdata
token = userdata.get('YOUR_AUTHTOKEN')

# To get your authtoken, sign up at https://dashboard.ngrok.com/signup and then go to https://dashboard.ngrok.com/get-started/your-authtoken
ngrok.set_auth_token(token)

public_url = ngrok.connect(8501)
print(f"Streamlit App URL: {public_url}")

# Run the Streamlit app in the background.
# This will print logs but the app will be accessible via the ngrok URL.
!streamlit run advanced_app.py &>/dev/null&
```
